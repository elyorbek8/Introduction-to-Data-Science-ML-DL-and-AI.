#Removing NaN values: ```.dropna()```

````python
import pandas as pd
````

````python
# Creating a df
df = pd.read_csv('https://raw.githubusercontent.com/anvarnarz/praktikum_datasets/refs/heads/main/automobile_data_nan.csv')
df.head()
````

````python
# cheking NaN values
df.isna()
````

````python
df.isna().sum()
````

````python
df.isna().sum() / len(df) * 100
````

````python
# number of rows
len(df)
````

````python
# removing rows with a nan value
len(df.dropna())
````

````python
# number of columns
len(df.columns)
````

````python
# removing columns with a nan value
df.dropna(axis=1)
````

````python
# number of columns after removal
len(df.dropna(axis=1).columns)
````

````python
# setting a threshold
df.dropna(axis=1, thresh=50)
````

````python
# removing nan values in regard to a particular column
df.dropna(subset=['price'])
````

````python
len(df.dropna(subset=['price']))
````
