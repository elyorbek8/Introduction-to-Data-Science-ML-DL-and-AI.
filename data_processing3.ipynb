# ```.map()``` method

````python
df = pd.read_csv('https://raw.githubusercontent.com/anvarnarz/praktikum_datasets/main/automobile_data.csv')
df.head()
````

````python
# printing unique values
df['num-of-cylinders'].unique()
````

````python
# matching the unappropriate values with appropriate ones
nums = {'four': 4, 'six': 6, 'five': 5, 'three': 3, 'twelve':12, 'two':2, 'eight':8}
df['num-of-cylinders'].map(nums)
df.head()
````

````python
# creating a column for the price in Uzbek currency
df['price_uzs'] = df['price'].map(lambda x: x * 12989)
df.head()
````

# Grouping values: ```pd.cut()``` and ```pd.qcut()```

````python
df = pd.read_csv('https://raw.githubusercontent.com/anvarnarz/praktikum_datasets/refs/heads/main/world_population.csv', usecols = ['country', 'pop2021', 'area'], index_col = 'country')
df.head()
````

````python
df.pop2021
````

````python
boxes = [0, 10, 100, 1000, 5000, 10000, 30000, 50000, 100_000, 500_000, 1000_000, 1_500_000]
````

````python
pop2021 = df.pop2021
pop2021
````

````python
groups = pd.cut(pop2021, bins=boxes)
groups
````

````python
groups.value_counts()
````

````python
pop2021.loc['India']
````

````python
df['groups'] = groups
````

````python
df
````

````python
df.loc['Falkland Islands']
````

````python
df.loc['China']
````

````python
# Grouping and giving labels
# boxes = [0, 10, 100, 1000, 5000, 10000, 30000, 50000, 100_000, 500_000, 1000_000, 1_500_000]
names = ['nano', 'micro', 'mini', 'small', 'medium', 'average', 'big', 'large', 'huge', 'giant', 'colossal']
groups = pd.cut(pop2021, bins=boxes, labels=names)
groups
````

````python
df['pop_group'] = groups
df
````

````python
df.loc['India']
````

````python
# grouping automatically
groups = pd.cut(pop2021, bins = 4)
groups.value_counts()
````

````python
# grouping equally
groups = pd.qcut(pop2021, q=4)
groups.value_counts()
````
