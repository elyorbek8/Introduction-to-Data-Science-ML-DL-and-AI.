# Grouping values: ```df.groupby()```

````python
import pandas as pd
import numpy as np
````

````python
df = pd.read_csv('https://raw.githubusercontent.com/anvarnarz/praktikum_datasets/refs/heads/main/automobile_data.csv')
df.head()
````

````python
df.company.value_counts()
````

````python
# grouping by a column
df.groupby('company').mean(numeric_only= True)
````

````python
df.groupby('company').price.mean()
````

````python
df.groupby('company').price.max()
````

````python
df.groupby('company').price.min()
````

````python
# grouping by several columns
df.groupby(['company', 'body-style']).mean(numeric_only= True)
````

# Grouping values and applying aggregated (multiple) functions

````python
df.columns
````

````python
df.groupby('company').agg({
    'price': [min, max, np.mean]
})
````

````python
numeric_cols = df.select_dtypes(include= 'number').columns
numeric_cols
````

````python
df.groupby('company')[numeric_cols].agg([min, max, np.mean])
````
