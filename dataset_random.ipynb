# Shuffling dataset values: ```np.random.permutation()``` and ```df.take()```
  When we collect data, we may arrange it in a prticular order.
  
  But sometimes for research we need random values.
  
  So we need to shaffle the values in the dataset.

````python
import numpy as np
import pandas as pd
````

````python
# uploading dataset and pre-processing it
df = pd.read_csv('https://raw.githubusercontent.com/anvarnarz/praktikum_datasets/refs/heads/main/housing_LR.csv')
df
````

````python
df.columns
````

````python
df.rename(columns={'Unnamed: 0': 'id'}, inplace= True)
````

````python
df.set_index('id', inplace= True)
````

````python
# shaffling values
rand_idx = np.random.permutation(len(df))
rand_idx
````

````python
df.take(rand_idx)
````

````python
df
````

# Taking a particular number of random values as a sample:

````python
df.sample(9) # not repetitive
````

````python
df_sample = df.sample(200, replace= True) # repetitive
df_sample
````
