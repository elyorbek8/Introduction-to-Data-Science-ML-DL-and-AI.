#Tokenization

```python
import nltk

from nltk.tokenize import word_tokenize

nltk.download('punkt_tab')
text = 'This is a sentince'
tokens = word_tokenize(text)
print(tokens)
```

#Named Entity Recoognition

```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
text = "Charles & Keith  has headqurter in Singapore. It is expected that it will open a new branch in Tashkent by 2028."
doc = nlp(text)
for ent in doc.ents:
    print(ent.text, ent.label_)
```

#Russian

```python
import spacy.cli
spacy.cli.download("ru_core_news_sm")
```

```python
nlp = spacy.load('ru_core_news_sm')
text = "Yandex имеет штаб-квартиру в Сингапуре. Ожидается, что к 2028 году откроется новый филиал в Ташкенте."
doc = nlp(text)
for ent in doc.ents:
    print(ent.text, ent.label_)
```
