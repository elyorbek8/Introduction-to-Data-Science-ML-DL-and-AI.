# Replacing nan values: ```.fillna()```

````python
import pandas as pd
import numpy as np
````

````python
df = pd.read_csv('https://raw.githubusercontent.com/anvarnarz/praktikum_datasets/refs/heads/main/automobile_data_nan.csv')
df.head()
````

````python
# replacing all nan values with a particular value
df.fillna(77)
````

````python
# replacing only specific column values with particular values respectively
df.fillna({'wheel-base': 95, 'horsepower': 102})
````

````python
# mean values
df.mean(numeric_only=True)
````

````python
# replacing nan values with mean values
df.fillna(df.mean(numeric_only=True))
````

````python
df
````

````python
# forward filling
df.fillna(method='ffill')
````

````python
# backward filling
df.fillna(method='bfill')
````
